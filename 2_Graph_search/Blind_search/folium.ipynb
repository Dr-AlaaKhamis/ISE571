### Ad dthe labels of the POIs

```python
import folium
from IPython.display import display

center = (26.3071, 50.1459)                # KFUPM
source_point = (26.3032, 50.1594)          # Dhahran Techno Valley
destination_point = (26.3142, 50.1476)     # KFUPM Mall

m = folium.Map(
    location=center,
    zoom_start=15,
    zoom_control=False,
    scrollWheelZoom=False,
    dragging=False
)

def add_labeled_marker(point, label, color, icon):
    folium.Marker(
        location=point,
        icon=folium.Icon(
            color=color,
            icon=icon,
            prefix="fa"
        ),
        tooltip=folium.Tooltip(
            label,
            permanent=True,
            direction="top",
            offset=(0, -30)
        ),
        popup=label
    ).add_to(m)

add_labeled_marker(
    source_point,
    "Dhahran Techno Valley",
    "red",
    "university"
)

add_labeled_marker(
    center,
    "KFUPM",
    "blue",
    "graduation-cap"
)

add_labeled_marker(
    destination_point,
    "KFUPM Mall",
    "green",
    "shopping-bag"
)

display(m)
```
